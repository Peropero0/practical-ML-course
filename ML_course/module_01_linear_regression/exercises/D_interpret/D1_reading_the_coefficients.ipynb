# D1 · Reading the coefficients: units, standardization and the logarithm

**Type**: Interpret · **Level**: ●●○ independent · **Time**: about 35 minutes · **core path ★**

## Objective

Answer questions about the effect of each variable in a salary model, showing that the largest coefficient
is not the most important variable, and translating the coefficients of a log-linear model into
percentages.

## Dataset

`ds.salaries()`: 600 employees with `years_experience`, `weekly_hours`, `big_city` (0/1), `degree` (0/1) and the
gross annual `salary` in euros. The generating process is hidden: the logarithm of the salary is linear in the
features.

## Educational purpose

Learn to **read** a model, not just to train it: what a coefficient means when the features have
different units, and what it means when the target is a logarithm. It is the kind of question you get in interviews and
in model reviews.

## Task

1. Train `raw = LinearRegression()` on `salary` with the features as they are. Save `raw_coefs` (a `Series` with the names as index) and `raw_most_influential`: the name of the variable with the largest coefficient **in absolute value**.
2. Standardize the features (mean 0, standard deviation 1), retrain and save `std_coefs` (euros per one standard deviation of change in the feature) and `std_most_influential`.
3. Train `log_model = LinearRegression()` on `np.log(salary)` with the original features. Save `log_coefs` and the exact percentage effect `effect_pct = (exp(coef) − 1) · 100` as a `Series`. Extract `degree_effect_pct` and `year_effect_pct`.
4. Compare two identical people (10 years of experience, 40 hours) except that `profile_a` lives in a big city and has a degree and `profile_b` does not. Compute `salary_ratio` as the ratio between the predicted salaries (with `np.exp` of the predictions of the log model) and `ratio_from_coefs` as `exp(coef_city + coef_degree)`.
5. Assign to `true_false` a dictionary `{"a": ..., "b": ..., "c": ...}` with `True` or `False` for the three statements:
   - **a**: the largest coefficient in absolute value of the model on the raw target always indicates the most important variable;
   - **b**: in the model on the logarithm, every extra year of experience raises the salary by about 3%;
   - **c**: in the model on the logarithm, the gain in euros due to the degree is the same for someone who earns 20,000 euros and for someone who earns 60,000.
