## Objective

Train trees of increasing depth, observe how the training accuracy rises up to 1.0 while the test one stalls, and read the **rule at the root** of a small tree.

## Dataset

`load_breast_cancer`: 569 tumors with 30 measurements; `y = 1` if malignant. Stratified 70/30 split.

## Educational purpose

See the overfitting of a tree: more depth means more leaves and a memorized training set, but after a few splits the test set no longer improves. And learn to read a tree as a set of `if... then` rules.

## Task

1. `table` (index: `"1"`, `"2"`, `"3"`, `"5"`, `"no limit"`; columns `train`, `test`, `leaves`) with `DecisionTreeClassifier(max_depth=..., random_state=0)`.
2. `gap` (training accuracy minus test accuracy for `max_depth=None`) and `gain_beyond_3` (best test accuracy among the depths beyond 3, that is 5 and no limit, minus that of depth 3).
3. Tree of depth 2: `root` (name of the variable of the first split, with `tree.tree_.feature[0]`) and `root_threshold` (`tree.tree_.threshold[0]`).
