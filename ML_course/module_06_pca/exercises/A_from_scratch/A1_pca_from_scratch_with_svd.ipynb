## Objective

Implement `fit`, `transform` and `inverse_transform` of a PCA with the SVD of the centered data, check that they match `sklearn.decomposition.PCA` (up to the sign), get the same variances with the eigenvalues of the covariance matrix and
show that the mean squared reconstruction error with `k` components is the sum of the **discarded** variances.

## Dataset

`load_wine` (178 wines, 13 measurements) standardized: `Z` (178 × 13).

## Educational purpose

Remove all mystery from PCA: it is a centering, an SVD and a matrix multiplication. Understand the role of the sign (arbitrary), the relationship between SVD and covariance and why "explained variance" and "reconstruction error" are the same thing.

## Task

1. `pca_fit(X, k)` → `(mean, components, variances)`: mean per column, the first `k` components (`k × p` matrix, one per row) and their variances $S^2/(n-1)$ (**only the first `k`**) from the SVD of `X - mean` (`np.linalg.svd(..., full_matrices=False)`).
2. `pca_transform(X, mean, components)` (the scores) and `pca_inverse(P, mean, components)` (the reconstruction in the original space).
3. Compare with `PCA(k)`: `equal_components` (the absolute values coincide), `equal_variances` and `equal_reconstruction` (the reconstruction does **not** depend on the sign). Also save `total_variance` (sum of the variances of all the columns of `Z`, with `ddof=1`) and check that it equals the sum of **all** the variances of the components (compute them with `k = 13`).
4. **Eigenvalues:** compute `eigenvalues` = decreasing eigenvalues of `np.cov(Z.T)` with `np.linalg.eigh` and check that they coincide with the variances from the SVD.
5. **Reconstruction error:** for `k = 3` compute `mse` (mean squared error over all the cells between `Z` and the reconstruction) and `theoretical_mse` (= sum of the discarded variances × `(n-1)/n` / `p`). They must coincide.
